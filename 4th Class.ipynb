```python
item = int(input("Enter the number of items: "))
unit_price = 10
total_purchase = item * unit_price

member = input("Are you a member? (yes/no): ")

if member.lower() == "yes":
    discount = total_purchase * 0.10
    print(f"Your Total Purchase is {total_purchase} & After 10% Member Discount: {total_purchase - discount}")

elif member.lower() == "no":
    discount = total_purchase * 0.05
    print(f"Your Total Purchase is {total_purchase} & After 5% Non-Member Discount: {total_purchase - discount}")

else:
    print("Invalid input. Please enter yes or no.")
```
