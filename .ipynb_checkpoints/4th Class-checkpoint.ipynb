# Conditional Statement 

```python
age =  int (input("Enter Your Age " ))
if age < 18 :
    print(" You are an adult " )
else : 
    print(" You are not an Adult " )
```

```python
bill = int (input("Please Enter Your Bill Amount " ))
if bill > 50000:
    discount = bill*.10
    print(f"Your Discount is :{discount}  and after discounter amount is {bill + discount}")
else:
    print("soryy ! Discount is not for you ")
```

```python
salary = int (input("Enter Your Salary "))
if salary >=600000 and salary <=120000:
    tax = salary*.05
elif salary >120000 and salary <=180000:
    tax = salary*.10
else:
    tax = salary*.15

print(f"Your salary is {salary}")
print(f"Your Tax is {tax}")
print(f"After tax amount is {salary - tax}")
```

```python
Unit = int(input("Enter the number of unit consumed"))
if Unit < 100:
    per_unit = 18
    bill_amount = Unit*per_unit
elif Unit  < 200:
    per_unit = 22
    bill_amount = Unit*per_unit  
elif Unit  < 300:
    per_unit = 28
    bill_amount = Unit*per_unit
else:
    per_unit = 32
    bill_amount = Unit*per_unit

print(f" Your Unit is {Unit} and Your Total Bill is : {bill_amount}")    
```
